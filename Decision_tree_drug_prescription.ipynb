```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn import metrics

%matplotlib inline

import warnings
warnings.filterwarnings('ignore')
```

```python
path = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/drug200.csv'
dt_data = pd.read_csv(path)

dt_data
```

```python
dt_data.info()
```

```python
label_encoder = LabelEncoder()

dt_data['Sex'] = label_encoder.fit_transform(dt_data['Sex'])
dt_data['BP'] = label_encoder.fit_transform(dt_data['BP'])
dt_data['Cholesterol'] = label_encoder.fit_transform(dt_data['Cholesterol'])
#dt_data['Drug'] = label_encoder.fit_transform(dt_data['Drug'])
dt_data
```

```python
dt_data.isnull().sum()
```

```python
custom_map = {'drugA': 0, 'drugB': 1, 'drugC': 2, 'drugX': 3, 'drugY': 4}

dt_data['Drug_num'] = dt_data['Drug'].map(custom_map)
dt_data
```

```python
dt_data.drop('Drug',axis=1).corr()['Drug_num']
```

```python
category_counts = dt_data['Drug'].value_counts()

plt.bar(category_counts.index, category_counts.values, color='blue')
plt.xlabel('Drug')
plt.ylabel('Count')
plt.title('Category distribution')
#plt.xticks(rotation=45)

plt.show()
```

```python
y = dt_data['Drug']
X = dt_data.drop(['Drug','Drug_num'], axis=1)
```

```python
X_trainset, X_testset, y_trainset, y_testset = train_test_split(X, y, test_size=0.3, random_state=32)
```

```python
drugTree = DecisionTreeClassifier(criterion="entropy", max_depth = 4)
```

```python
drugTree.fit(X_trainset,y_trainset)
```

```python
tree_predictions = drugTree.predict(X_testset)
```

```python
print("Decision Trees's Accuracy: ", metrics.accuracy_score(y_testset, tree_predictions))
```

```python
plot_tree(drugTree)
plt.show()
```
