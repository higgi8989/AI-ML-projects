```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
%matplotlib inline
```

```python
#import the csv file data and convert it into pandas dataframe
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/teleCust1000t.csv"
df = pd.read_csv(url)
df.head()
```

```python
# seprating target values and training values
X = df.drop('custcat', axis = 1)
y = df['custcat'].astype('category')
```

```python
#normalizing the dataset
X_norm = StandardScaler().fit_transform(X)
```

```python
#spliting the training and testing data
X_train, X_test, y_train, y_test = train_test_split(X_norm, y, test_size = 0.2, random_state = 32)
```

```python
# training the model
k = 20
#Train Model and Predict  
knn_classifier = KNeighborsClassifier(n_neighbors=k)
knn_model = knn_classifier.fit(X_train,y_train)
"""Ks = 100
acc = np.zeros((Ks))
std_acc = np.zeros((Ks))
for n in range(1,Ks+1):
    #Train Model and Predict  
    knn_model_n = KNeighborsClassifier(n_neighbors = n).fit(X_train,y_train)
    y_prep = knn_model_n.predict(X_test)
    acc[n-1] = accuracy_score(y_test, y_prep)
    std_acc[n-1] = np.std(y_prep==y_test)/np.sqrt(y_prep.shape[0])"""
```
